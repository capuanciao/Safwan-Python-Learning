```python
import numpy as np
amplitudes = np.array ([0.2, -0.4, 1.1, 0.7, -0.1])

print (amplitudes)
print ("Number of samples:"), len(amplitudes)
print ("Average amplitude;", np.mean(amplitudes))

```

```python
scaled = amplitudes * 2
print (scaled)
```

```python
time_ms = np.arange(0, 100, 2)

trace = np.sin(2 * np.pi * time_ms / 40)

print("Number of samples:", len(trace))
print("First five amplitudes:", trace[:5])
```
